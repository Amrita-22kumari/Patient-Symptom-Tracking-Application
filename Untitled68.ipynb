Create patient symptom data model

```python
%%writefile /content/patient-symptom-tracker/backend/models.py

from sqlalchemy import Column, Integer, String, DateTime
from datetime import datetime, timezone

from database import Base


class SymptomLog(Base):

    __tablename__ = "symptom_logs"

    id = Column(
        Integer,
        primary_key=True,
        index=True
    )

    patient_id = Column(
        String(100),
        nullable=False,
        index=True
    )

    symptom_description = Column(
        String(500),
        nullable=False
    )

    severity = Column(
        Integer,
        nullable=False
    )

    timestamp = Column(
        DateTime,
        nullable=False,
        unique=True,
        index=True,
        default=lambda: datetime.now(
            timezone.utc
        ).replace(tzinfo=None)
    )
```

Create input validation

```python
%%writefile /content/patient-symptom-tracker/backend/schemas.py

from pydantic import BaseModel, Field, field_validator
from datetime import datetime


class SymptomCreate(BaseModel):

    patient_id: str = Field(
        min_length=1,
        max_length=100
    )

    symptom_description: str = Field(
        min_length=1,
        max_length=500
    )

    severity: int = Field(
        ge=1,
        le=10
    )

    @field_validator(
        "patient_id",
        "symptom_description"
    )
    @classmethod
    def validate_not_empty(cls, value):

        if not value.strip():
            raise ValueError("Field cannot be empty")

        return value.strip()


class SymptomResponse(BaseModel):

    id: int
    patient_id: str
    symptom_description: str
    severity: int
    timestamp: datetime

    class Config:
        from_attributes = True
```

Backend development (Create CRUD operations)

```python
%%writefile /content/patient-symptom-tracker/backend/crud.py

from sqlalchemy.orm import Session
from sqlalchemy import func

from datetime import datetime, timedelta, timezone

from models import SymptomLog


def create_symptom(db: Session, symptom):

    latest = db.query(
        func.max(SymptomLog.timestamp)
    ).scalar()

    current_time = datetime.now(
        timezone.utc
    ).replace(tzinfo=None)

    if latest is not None and current_time <= latest:
        current_time = latest + timedelta(microseconds=1)

    new_symptom = SymptomLog(
        patient_id=symptom.patient_id,
        symptom_description=symptom.symptom_description,
        severity=symptom.severity,
        timestamp=current_time
    )

    db.add(new_symptom)
    db.commit()
    db.refresh(new_symptom)

    return new_symptom


def get_patient_symptoms(db: Session, patient_id):

    return (
        db.query(SymptomLog)
        .filter(SymptomLog.patient_id == patient_id)
        .order_by(
            SymptomLog.timestamp.asc(),
            SymptomLog.id.asc()
        )
        .all()
    )


def get_symptom_trends(db: Session, patient_id):

    cutoff = (
        datetime.now(timezone.utc).replace(tzinfo=None)
        - timedelta(days=30)
    )

    logs = (
        db.query(SymptomLog)
        .filter(
            SymptomLog.patient_id == patient_id,
            SymptomLog.timestamp >= cutoff
        )
        .order_by(SymptomLog.timestamp.asc())
        .all()
    )

    daily_data = {}

    for log in logs:

        day = log.timestamp.date().isoformat()

        daily_data.setdefault(day, []).append(
            log.severity
        )

    trends = []

    for day, values in sorted(daily_data.items()):

        trends.append({
            "date": day,
            "average_severity": round(
                sum(values) / len(values), 2
            )
        })

    return trends
```

Create FastAPI application

```python
%%writefile /content/patient-symptom-tracker/backend/main.py

from fastapi import FastAPI, Depends, HTTPException
from fastapi.staticfiles import StaticFiles
from fastapi.responses import FileResponse
from sqlalchemy.orm import Session
from pathlib import Path

import database
import crud
import schemas

from database import get_db

database.Base.metadata.create_all(
    bind=database.engine
)

app = FastAPI(
    title="Patient Symptom Tracker",
    version="1.0.0"
)

BASE_DIR = Path(__file__).resolve().parent.parent

FRONTEND_DIR = BASE_DIR / "frontend"


@app.get("/")
def home():

    return FileResponse(
        FRONTEND_DIR / "index.html"
    )


app.mount(
    "/static",
    StaticFiles(directory=str(FRONTEND_DIR)),
    name="static"
)


@app.post(
    "/symptoms",
    response_model=schemas.SymptomResponse
)
def submit_symptom(
    symptom: schemas.SymptomCreate,
    db: Session = Depends(get_db)
):

    try:

        result = crud.create_symptom(db, symptom)

        return result

    except Exception:

        db.rollback()

        raise HTTPException(
            status_code=500,
            detail="Unable to save symptom"
        )


@app.get(
    "/patients/{patient_id}/symptoms",
    response_model=list[schemas.SymptomResponse]
)
def get_symptoms(
    patient_id: str,
    db: Session = Depends(get_db)
):

    if not patient_id.strip():

        raise HTTPException(
            status_code=400,
            detail="Patient ID cannot be empty"
        )

    return crud.get_patient_symptoms(
        db,
        patient_id
    )


@app.get(
    "/patients/{patient_id}/symptoms/trends"
)
def symptom_trends(
    patient_id: str,
    db: Session = Depends(get_db)
):

    if not patient_id.strip():

        raise HTTPException(
            status_code=400,
            detail="Patient ID cannot be empty"
        )

    return crud.get_symptom_trends(
        db,
        patient_id
    )


@app.get("/health")
def health_check():

    return {
        "status": "healthy"
    }
```

Creating JavaScript

```python
%%writefile /content/patient-symptom-tracker/frontend/script.js

const form = document.getElementById("symptomForm");

const message = document.getElementById("message");

const button = document.getElementById("submitButton");

let severityChart = null;


function showMessage(text, type) {

    message.textContent = text;

    message.className = type;

}


form.addEventListener("submit", async function(event) {

    event.preventDefault();

    const patientId =
        document.getElementById("patientId").value.trim();

    const description =
        document.getElementById("description").value.trim();

    const severityValue =
        document.getElementById("severity").value;

    const severity = Number(severityValue);


    if (!patientId || !description || !severityValue) {

        showMessage("Please complete all fields.", "error");

        return;

    }


    if (!Number.isInteger(severity) ||
        severity < 1 || severity > 10) {

        showMessage("Severity must be between 1 and 10.", "error");

        return;

    }


    button.disabled = true;


    try {

        const response = await fetch("/symptoms", {

            method: "POST",

            headers: {
                "Content-Type": "application/json"
            },

            body: JSON.stringify({
                patient_id: patientId,
                symptom_description: description,
                severity: severity
            })

        });


        const data = await response.json();


        if (!response.ok) {

            throw new Error(
                typeof data.detail === "string"
                    ? data.detail
                    : JSON.stringify(data.detail)
            );

        }


        showMessage("Symptom saved successfully!", "success");

        document.getElementById("description").value = "";

        document.getElementById("severity").value = "";

        document.getElementById("searchPatient").value = patientId;

        await loadSymptoms(patientId);


    } catch (error) {

        showMessage(
            "Submission failed: " + error.message,
            "error"
        );

    } finally {

        button.disabled = false;

    }

});


async function loadSymptoms(patientId = null) {

    patientId = patientId ||
        document.getElementById("searchPatient").value.trim();

    const table = document.getElementById("symptomTable");

    if (!patientId) {

        showMessage("Enter patient ID.", "error");

        return;

    }

    table.innerHTML = "";

    try {

        const response = await fetch(
            `/patients/${encodeURIComponent(patientId)}/symptoms`
        );

        if (!response.ok) {
            throw new Error("Unable to retrieve records.");
        }

        const symptoms = await response.json();

        symptoms.forEach(symptom => {

            const row = document.createElement("tr");

            const values = [
                symptom.id,
                symptom.symptom_description,
                symptom.severity,
                new Date(symptom.timestamp + "Z").toLocaleString()
            ];

            values.forEach(value => {

                const cell = document.createElement("td");

                cell.textContent = value;

                row.appendChild(cell);

            });

            table.appendChild(row);

        });

        document.getElementById("recordCount").textContent =
            "Total records: " + symptoms.length;

        showMessage("Records loaded successfully.", "success");

        await loadTrends(patientId);

    } catch (error) {

        showMessage(
            "Unable to load records: " + error.message,
            "error"
        );

    }

}


async function loadTrends(patientId) {

    try {

        const response = await fetch(
            `/patients/${encodeURIComponent(patientId)}/symptoms/trends`
        );

        if (!response.ok) {
            throw new Error("Unable to load trends.");
        }

        const trends = await response.json();

        const labels = trends.map(item => item.date);

        const values = trends.map(
            item => item.average_severity
        );

        if (severityChart) {
            severityChart.destroy();
        }

        severityChart = new Chart(
            document.getElementById("severityChart"),
            {
                type: "line",

                data: {
                    labels: labels,

                    datasets: [{
                        label: "Average Severity",
                        data: values,
                        borderColor: "#2456a6",
                        backgroundColor: "rgba(36,86,166,0.15)",
                        fill: true,
                        tension: 0.3
                    }]
                },

                options: {
                    responsive: true,

                    scales: {
                        y: {
                            min: 0,
                            max: 10
                        }
                    }
                }
            }
        );

        document.getElementById("chartMessage").textContent =
            trends.length
                ? "Showing daily averages for the last 30 days."
                : "No symptom data available for the last 30 days.";

    } catch (error) {

        document.getElementById("chartMessage").textContent =
            error.message;

    }

}
```

```python
%%writefile /content/patient-symptom-tracker/tests/test_api.py

import sys
import pytest

from datetime import datetime
from sqlalchemy import create_engine
from sqlalchemy.orm import sessionmaker
from sqlalchemy.pool import StaticPool
from fastapi.testclient import TestClient

sys.path.insert(
    0,
    "/content/patient-symptom-tracker/backend"
)

from main import app
from database import Base, get_db


# Create an isolated in-memory test database
SQLALCHEMY_DATABASE_URL = "sqlite://"

engine = create_engine(
    SQLALCHEMY_DATABASE_URL,
    connect_args={"check_same_thread": False},
    poolclass=StaticPool
)

TestingSessionLocal = sessionmaker(
    autocommit=False,
    autoflush=False,
    bind=engine
)


def override_get_db():
    Base.metadata.create_all(bind=engine)

    db = TestingSessionLocal()

    try:
        yield db
    finally:
        db.close()


app.dependency_overrides[get_db] = override_get_db

client = TestClient(app)


@pytest.fixture(autouse=True)
def reset_database():
    Base.metadata.drop_all(bind=engine)
    Base.metadata.create_all(bind=engine)


def test_create_symptom():

    response = client.post(
        "/symptoms",
        json={
            "patient_id": "P001",
            "symptom_description": "Headache",
            "severity": 5
        }
    )

    assert response.status_code == 200
    assert response.json()["patient_id"] == "P001"


def test_reject_empty_symptom():

    response = client.post(
        "/symptoms",
        json={
            "patient_id": "P001",
            "symptom_description": "   ",
            "severity": 5
        }
    )

    assert response.status_code == 422


@pytest.mark.parametrize("severity", [0, 11, -1, 15])
def test_reject_invalid_severity(severity):

    response = client.post(
        "/symptoms",
        json={
            "patient_id": "P001",
            "symptom_description": "Headache",
            "severity": severity
        }
    )

    assert response.status_code == 422


def test_store_five_symptoms():

    symptoms = [
        ("Headache", 5),
        ("Fever", 7),
        ("Cough", 4),
        ("Fatigue", 6),
        ("Body pain", 8)
    ]

    for description, severity in symptoms:

        response = client.post(
            "/symptoms",
            json={
                "patient_id": "P001",
                "symptom_description": description,
                "severity": severity
            }
        )

        assert response.status_code == 200

    response = client.get(
        "/patients/P001/symptoms"
    )

    assert response.status_code == 200

    records = response.json()

    assert len(records) >= 5

    timestamps = [
        record["timestamp"]
        for record in records
    ]

    assert len(timestamps) == len(set(timestamps))

    assert timestamps == sorted(timestamps)


def test_patient_records_are_separate():

    client.post(
        "/symptoms",
        json={
            "patient_id": "P001",
            "symptom_description": "Headache",
            "severity": 5
        }
    )

    client.post(
        "/symptoms",
        json={
            "patient_id": "P002",
            "symptom_description": "Fever",
            "severity": 7
        }
    )

    response = client.get(
        "/patients/P001/symptoms"
    )

    records = response.json()

    assert all(
        record["patient_id"] == "P001"
        for record in records
    )
```
